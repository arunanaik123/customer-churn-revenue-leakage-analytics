```python
# ============================================================
# SETUP: Load the model's predictions saved from 03_churn_model.ipynb
# ============================================================
import pandas as pd

results = pd.read_csv('../data/processed/model_predictions.csv')
print(results.shape)
results.head()
```

```python
# ============================================================
# 1. TOTAL REVENUE AT RISK — everyone who actually churned
#    (in this test set, not the whole original dataset)
# ============================================================
actual_churners = results[results['actual_churn'] == 1]
total_revenue_at_risk = actual_churners['MonthlyCharges_actual'].sum()

print(f"Customers who actually churned (test set): {len(actual_churners)}")
print(f"Total monthly revenue lost: ${total_revenue_at_risk:,.2f}")
print(f"Annualized: ${total_revenue_at_risk * 12:,.2f}")
```

```python
# ============================================================
# 2. REVENUE THE MODEL SUCCESSFULLY CAUGHT
#    (customers who churned AND the model correctly flagged them)
# ============================================================
caught = results[(results['actual_churn'] == 1) & (results['predicted_churn'] == 1)]
revenue_caught = caught['MonthlyCharges_actual'].sum()

print(f"Churners the model caught: {len(caught)}")
print(f"Revenue the model could have helped save: ${revenue_caught:,.2f}")
print(f"% of total at-risk revenue caught: {round(revenue_caught/total_revenue_at_risk*100, 1)}%")
```

```python
# ============================================================
# 3. REVENUE THE MODEL MISSED
#    (customers who churned but the model did NOT flag them)
# ============================================================
missed = results[(results['actual_churn'] == 1) & (results['predicted_churn'] == 0)]
revenue_missed = missed['MonthlyCharges_actual'].sum()

print(f"Churners the model missed: {len(missed)}")
print(f"Revenue still lost with no warning: ${revenue_missed:,.2f}")
```

```python
# ============================================================
# 4. COST OF FALSE ALARMS 
#    (Customers the model flagged as risky, but who actually stayed)
# ============================================================
false_alarms = results[(results['actual_churn'] == 0) & (results['predicted_churn'] == 1)]
print(f"False alarms: {len(false_alarms)} customers flagged unnecessarily")
```
